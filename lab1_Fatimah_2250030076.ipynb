## Mini Project: Simple Dataset

Objective: Perform basic analysis on the `study_hours` and `marks` dataset.

```python
# Display the dataset
print('Study Hours:', study_hours)
print('Marks Achieved:', marks)

print('\n--- Study Hours Analysis ---\n')
print(f'Number of observations (study hours): {len(study_hours)}')
print(f'Total study hours: {sum(study_hours)}')
print(f'Average study hours: {sum(study_hours) / len(study_hours):.2f}')
print(f'Minimum study hours: {min(study_hours)}')
print(f'Maximum study hours: {max(study_hours)}')

print('\n--- Marks Achieved Analysis ---\n')
print(f'Number of observations (marks): {len(marks)}')
print(f'Total marks: {sum(marks)}')
print(f'Average marks: {sum(marks) / len(marks):.2f}')
print(f'Minimum mark: {min(marks)}')
print(f'Maximum mark: {max(marks)}')

# Analyze relationship and pass/fail status
print('\n--- Individual Sample Analysis ---\n')
passing_students_count = 0
for hours, mark in zip(study_hours, marks):
    status = "Pass" if mark >= 60 else "Fail"
    print(f'Studied for {hours} hours, achieved {mark} marks. Status: {status}')
    if mark >= 60:
        passing_students_count += 1

print(f'\nTotal students who achieved a mark of 60 or above: {passing_students_count}')
```

6. Why is it useful to store data in a structured form before applying a Machine Learning algorithm?

Storing data in a structured form (like lists, or more commonly, DataFrames in pandas) is crucial for Machine Learning because:

1- Organization and Accessibility: Structured data is easy to organize, understand, and access programmatically. Each piece of information has a defined place.
Data Consistency: It helps maintain data consistency and integrity, reducing errors and ambiguities.
2- Algorithm Compatibility: Most machine learning algorithms are designed to work with structured data. They expect inputs in a tabular format (rows and columns) to identify patterns and relationships.
3- Preprocessing: Structured data simplifies preprocessing steps like handling missing values, scaling, and encoding, which are essential before training a model.
4- Interpretability: It makes the data and model's behavior more interpretable, as you can clearly see which features (columns) are influencing the predictions.
