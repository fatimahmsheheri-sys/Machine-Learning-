```python
def perform_arithmetic_operation(num1, num2, operation):
    if operation == '+':
        return num1 + num2
    elif operation == '-':
        return num1 - num2
    elif operation == '*':
        return num1 * num2
    elif operation == '/':
        if num2 == 0:
            return "Error: Division by zero"
        return num1 / num2
    elif operation == '//':
        if num2 == 0:
            return "Error: Division by zero"
        return num1 // num2
    elif operation == '%':
        if num2 == 0:
            return "Error: Modulo by zero"
        return num1 % num2
    elif operation == '**':
        return num1 ** num2
    else:
        return "Error: Invalid operation"

# Demonstration of the function
x = 25
y = 5

print(f"Addition ({x} + {y}):", perform_arithmetic_operation(x, y, '+'))
print(f"Subtraction ({x} - {y}):", perform_arithmetic_operation(x, y, '-'))
print(f"Multiplication ({x} * {y}):", perform_arithmetic_operation(x, y, '*'))
print(f"Division ({x} / {y}):", perform_arithmetic_operation(x, y, '/'))
print(f"Floor Division ({x} // {y}):", perform_arithmetic_operation(x, y, '//'))
print(f"Remainder ({x} % {y}):", perform_arithmetic_operation(x, y, '%'))
print(f"Power ({x} ** {y}):", perform_arithmetic_operation(x, y, '**'))

# Example of invalid operation
print("Invalid operation:", perform_arithmetic_operation(x, y, '&'))
# Example of division by zero
print("Division by zero:", perform_arithmetic_operation(x, 0, '/'))
```

## Mini Project: Simple Dataset

Objective: Perform basic analysis on the `study_hours` and `marks` dataset.

```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

# Display the dataset
print('Study Hours:', study_hours)
print('Marks Achieved:', marks)

print('\n--- Study Hours Analysis ---\n')
print(f'Number of observations (study hours): {len(study_hours)}')
print(f'Total study hours: {sum(study_hours)}')
print(f'Average study hours: {sum(study_hours) / len(study_hours):.2f}')
print(f'Minimum study hours: {min(study_hours)}')
print(f'Maximum study hours: {max(study_hours)}')

print('\n--- Marks Achieved Analysis ---\n')
print(f'Number of observations (marks): {len(marks)}')
print(f'Total marks: {sum(marks)}')
print(f'Average marks: {sum(marks) / len(marks):.2f}')
print(f'Minimum mark: {min(marks)}')
print(f'Maximum mark: {max(marks)}')

# Analyze relationship and pass/fail status
print('\n--- Individual Sample Analysis ---\n')
passing_students_count = 0
for hours, mark in zip(study_hours, marks):
    status = "Pass" if mark >= 60 else "Fail"
    print(f'Studied for {hours} hours, achieved {mark} marks. Status: {status}')
    if mark >= 60:
        passing_students_count += 1

print(f'\nTotal students who achieved a mark of 60 or above: {passing_students_count}')
```
